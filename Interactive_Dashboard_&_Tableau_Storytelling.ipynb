Q1. Explain the difference between a Tableau Dashboard and a Tableau Story Point. Give two healthcare-related use cases.

-> A Tableau Dashboard is a collection of multiple visualizations displayed on a single screen to analyze data interactively. It helps users compare and monitor different metrics together.

A Tableau Story Point is a sequence of visualizations arranged to explain insights step-by-step like a presentation. It helps in storytelling using data

Difference:

| Dashboard                           | Story Point                            |
| ----------------------------------- | -------------------------------------- |
| Shows multiple charts on one screen | Shows insights step-by-step            |
| Used for interactive analysis       | Used for presentation and storytelling |
| Best for monitoring KPIs            | Best for explaining findings           |

Healthcare Use Cases:

- A hospital dashboard to monitor patient count, department cost, and patient outcomes in real time.
- A story point presentation to explain treatment trends and doctor performance to hospital management.

Q2. What are Actions in Tableau? Explain how Filter Action can help hospital management analyze patient data.

-> Actions in Tableau are interactive features that allow users to interact with visualizations dynamically. Common actions include Filter Action, Highlight Action, and URL Action.

Filter Action helps hospital management by filtering related data when a user clicks on a chart. For example, if management clicks on the “North Region” in a patient chart, Tableau will display only patients, treatment costs, and department details for that region. This helps in better decision-making and focused analysis.
